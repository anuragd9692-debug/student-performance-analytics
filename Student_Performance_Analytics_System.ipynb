```python
# Step 1:Importing Libraries & Creating the dataSet
import pandas as pd
import numpy as np

data = {
    "Student_ID": list(range(101, 121)),
    "Name": [
        "Rahul", "Priya", "Amit", "Sneha", "Rohan",
        "Anjali", "Vikas", "Neha", "Arjun", "Pooja",
        "Karan", "Riya", "Sumit", "Kavya", "Manish",
        "Divya", "Raj", "Simran", "Akash", "Nisha"
    ],
    "Department": [
        "CS", "IT", "CS", "IT", "CS",
        "IT", "CS", "IT", "CS", "IT",
        "CS", "IT", "CS", "IT", "CS",
        "IT", "CS", "IT", "CS", "IT"
    ],
    "Python": [
        85, 76, 92, 45, 67, 88, 32, 79, 95, 56,
        72, 90, 41, 83, 61, 74, 87, 68, 93, 80
    ],
    "Mathematics": [
        78, 82, 95, 52, 71, 84, 40, 75, 91, 60,
        68, 89, 48, 86, 64, 79, 90, 73, 96, 85
    ],
    "English": [
        80, 74, 91, 61, 69, 90, 55, 81, 94, 58,
        75, 92, 44, 79, 66, 82, 86, 77, 89, 88
    ],
    "Attendance": [
        90, 85, 95, 72, 80, 93, 65, 88, 98, 76,
        82, 96, 70, 91, 84, 87, 92, 81, 97, 94
    ]
}

pd.DataFrame(data).to_csv("students.csv", index=False)

print("Dataset created successfully.")
```

```python
# Reading & check the CsV file :
df = pd.read_csv("students.csv")

print("Total records:", len(df))
print("Columns:", df.columns.tolist())

print("\nDataset information:")
# Checking the info:
df.info()

print("\nFirst five records:")
# Checking the head:
display(df.head())
```

```python
# Step 3 : Calculating total & average marks
subjects = ["Python", "Mathematics", "English"]

df["Total"] = df[subjects].sum(axis=1)
df["Average"] = np.round(
    df[subjects].mean(axis=1), 2
)

display(df[["Name", "Total", "Average"]])
```

```python
# Step 4 : Creating a grade-function
def assign_grade(average):
    if average >= 90:
        return "A+"
    elif average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    elif average >= 40:
        return "D"
    else:
        return "F"

df["Grade"] = df["Average"].apply(assign_grade)

display(df[["Name", "Average", "Grade"]])
```

```python
# Step 5: Check Pass or Fail
def check_result(row):
    for subject in subjects:
        if row[subject] < 35:
            return "Fail"
    return "Pass"

df["Result"] = df.apply(check_result, axis=1)

display(df[["Name", "Python", "Mathematics",
            "English", "Result"]])
```

```python
# Step 6 : Analyzing Overall Class Performance
print("CLASS PERFORMANCE REPORT")
print("Total students:", len(df))
print("Class average:", round(np.mean(df["Average"]), 2))
print("Highest average:", df["Average"].max())
print("Lowest average:", df["Average"].min())
print("Students passed:", (df["Result"] == "Pass").sum())
print("Students failed:", (df["Result"] == "Fail").sum())
```

```python
# Step 7 : Performance Subject-wise Analysis (Using a Loop)
print("SUBJECT-WISE PERFORMANCE")
for subject in subjects:
    print(
        subject,
        "average:",
        round(np.mean(df[subject]), 2)
    )

subject_averages = df[subjects].mean()
best_subject = subject_averages.idxmax()

print("\nBest-performing subject:", best_subject)
```

```python
# Step 8 : Finding the Top performing Students
highest_average = df["Average"].max()

top_students = df[
    df["Average"] == highest_average
]

print("TOP-PERFORMING STUDENTS")
display(top_students[["Name", "Department", "Average", "Grade"]])
```

```python
# Step 9 : Analyzing the attendance's
print("Average attendance:",
      round(df["Attendance"].mean(), 2), "%")

print("\nStudents with attendance below 75%:")

display(
    df.loc[
        df["Attendance"] < 75,
        ["Name", "Attendance"]
    ]
)
```

```python
# Step 10: Display the complete final report of student's
print("STUDENT PERFORMANCE ANALYTICS SYSTEM")
display(
    df[[
        "Student_ID", "Name", "Department",
        "Python", "Mathematics", "English",
        "Total", "Average", "Grade",
        "Result", "Attendance"
    ]]
)
```

```python
# Step 11 : Save the final analysis
df.to_csv("analysis_results.csv", index=False)

print("Final analysis saved to analysis_results.csv")
```
